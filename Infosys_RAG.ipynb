```python
pages = [(i + 1, p.extract_text() or '') for i, p in enumerate(reader.pages)]
print(len(pages))
```

```python
chunk_size = 1000
overlap = 200
chunks = []

for page_num, text in pages:
    start = 0
    while start < len(text):
        piece = text[start:start + chunk_size]
        if len(piece.strip()) > 50:
            chunks.append({'page': page_num, 'text': piece})
        start += chunk_size - overlap

print(len(chunks))
```

```python
from sentence_transformers import SentenceTransformer
embed_model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
def search(question, k=3):
    q = embed_model.encode([question], normalize_embeddings=True)
    scores = embeddings @ q[0]
    top = scores.argsort()[::-1][:k]
    return [(chunks[i]['page'], round(float(scores[i]), 3), chunks[i]['text']) for i in top]
```

```python
texts = [c['text'] for c in chunks]
embeddings = embed_model.encode(texts, show_progress_bar=True, normalize_embeddings=True)
print(embeddings.shape)
```

```python
def ask(question, k=5):
    results = search(question, k)
    context = ''
    for page, score, text in results:
        context += f'[PDF page {page}]\n{text}\n\n'
    prompt = f"""You answer questions about the Infosys Integrated Annual Report 2025-26.
Use ONLY the context below. Cite the PDF page numbers you used, like (page 72).
If the answer is not in the context, say exactly: I could not find this in the report.

Context:
{context}
Question: {question}
Answer:"""
    response = client.models.generate_content(model=MODEL, contents=prompt)
    return response.text, [r[0] for r in results]
```

```python
import re, time
import pandas as pd

test_set = [
    ('What was the revenue of Infosys in FY2026 in rupees crore?', ['178650'], [16]),
    ('What was the net profit attributable to owners of the company in FY2026?', ['29440'], [16]),
    ('What was the basic earnings per share in FY2026?', ['71.58'], [16]),
    ('What was the adjusted operating margin in FY2026?', ['21.0', '21%'], [16]),
    ('What was the total dividend per share for FY2026?', ['48'], [16, 38]),
    ('How large was the buyback completed in FY2026?', ['18000'], [16, 38]),
    ('What was the return on equity?', ['31.6'], [16]),
    ('What was the free cash flow in FY2026?', ['33097'], [16]),
    ('How many US$50 million plus clients does Infosys have?', ['88'], [16]),
    ('For how many years in a row has Infosys been carbon neutral?', ['7years', 'seven'], [16]),
    ('What share of India operations is powered by renewable energy?', ['81.8'], [16]),
    ('Who is the CEO and Managing Director of Infosys?', ['salilparekh'], [11, 54]),
    ('Who is the Chairman of Infosys?', ['nandan'], [11]),
    ('Who was appointed Vice Chairman effective April 30, 2026?', ['paranjpe'], [11, 13]),
    ('In how many countries does Infosys do business?', ['59'], [10]),
    ('When will the Annual General Meeting be held?', ['june23'], [38]),
    ('What is the record date for the final dividend?', ['june10'], [38]),
]

not_in_report = [
    "What is the CEO's favourite food?",
    "What was TCS's revenue in FY2026?",
    "What is the Infosys share price today?",
    "Who won the 2026 FIFA World Cup?",
]
```

```python
def norm(s):
    s = re.sub(r'\((?:pdf )?pages?[^)]*\)', '', s, flags=re.I)
    return re.sub(r'[\s,₹`]', '', s.lower())

def safe_ask(q):
    try:
        return ask(q)
    except Exception:
        time.sleep(30)
        return ask(q)

rows = []
for q, keys, pages in test_set:
    answer, got = safe_ask(q)
    rows.append({'question': q,
                 'retrieved_ok': any(p in got for p in pages),
                 'answer_ok': any(k in norm(answer) for k in keys),
                 'answer': answer})
    time.sleep(8)

refusals = []
for q in not_in_report:
    answer, got = safe_ask(q)
    refusals.append({'question': q, 'refused': 'could not find' in answer.lower(), 'answer': answer})
    time.sleep(8)

df_eval = pd.DataFrame(rows)
df_ref = pd.DataFrame(refusals)
print('Retrieval hit rate:', round(df_eval['retrieved_ok'].mean(), 2))
print('Answer accuracy:', round(df_eval['answer_ok'].mean(), 2))
print('Correct refusals:', int(df_ref['refused'].sum()), 'of', len(df_ref))
```

```python
from rank_bm25 import BM25Okapi
import numpy as np, re

def tokenize(t):
    return re.findall(r'\w+', t.lower())

bm25 = BM25Okapi([tokenize(c['text']) for c in chunks])

def hybrid_search(question, k=5, pool=30):
    q = embed_model.encode([question], normalize_embeddings=True)
    sem = embeddings @ q[0]
    kw = bm25.get_scores(tokenize(question))
    sem_rank = sem.argsort()[::-1][:pool]
    kw_rank = np.argsort(kw)[::-1][:pool]
    fused = {}
    for r, i in enumerate(sem_rank):
        fused[i] = fused.get(i, 0) + 1 / (60 + r)
    for r, i in enumerate(kw_rank):
        fused[i] = fused.get(i, 0) + 1 / (60 + r)
    top = sorted(fused, key=fused.get, reverse=True)[:k]
    return [(chunks[i]['page'], round(float(fused[i]), 4), chunks[i]['text']) for i in top]
```

```python
results = [baseline, hybrid]

def run_eval(label, k=5):
    rows = []
    for q, keys, pages in test_set:
        try:
            answer, got = ask(q, k)
        except Exception:
            time.sleep(30)
            answer, got = ask(q, k)
        rows.append({'question': q,
                     'retrieved_ok': any(p in got for p in pages),
                     'answer_ok': any(key in norm(answer) for key in keys),
                     'answer': answer})
        time.sleep(8)
    refusals = []
    for q in not_in_report:
        try:
            answer, got = ask(q, k)
        except Exception:
            time.sleep(30)
            answer, got = ask(q, k)
        refusals.append({'refused': 'could not find' in answer.lower()})
        time.sleep(8)
    d = pd.DataFrame(rows)
    r = pd.DataFrame(refusals)
    results.append({'method': label,
                    'retrieval_hit': round(d['retrieved_ok'].mean(), 2),
                    'answer_acc': round(d['answer_ok'].mean(), 2),
                    'refusals': f"{int(r['refused'].sum())}/{len(r)}"})
    return d
```
